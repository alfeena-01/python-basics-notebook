REMOVE LIST ITEMS 
 1. remove()
 2. pop()
 3. del keyword

Remove specified item
the remove() method removes the specified item. 

```python
thislist=['apple','banana','cherry']
thislist.remove('banana')
print(thislist)

```

If there are more than one item with the specified value, the remove() method removes the first occurrence:   


```python
##mremoves the first occurance of "banana"

thislist = ['apple','banana','cherry','banana', 'kiwi']
thislist.remove('banana')
print(thislist)
```

REMOVES SPECIFIES INDEX

The pop() method removes the specified index:

```python
thislist = ['apple','banana','cherry','banana', 'kiwi']
thislist.pop(1)
print(thislist)
```

If we do not specify index , the pop() method removes the last item .

```python
##remove last item 

thislist = ['apple','banana','cherry','banana', 'kiwi']
thislist.pop()
print(thislist)
```

the del keyword also removes the specifeid index:


```python
thislist = ['apple','banana','cherry','banana', 'kiwi']
del.thislist[0]
print(thislist)
```

SORT LIST ALPHANUMERICALLY

list objects have a sort() method that will sort the lsit alphanumerically.
ascending by default:

```python
##sort the lsit alphabetically
thislist = ['apple','banana','cherry','banana', 'kiwi']
thislist.sort(reverse =True) #as discending
print(thislist)

# sort() is ascending
thislist = ['apple','banana','cherry','banana', 'kiwi']
thislist.sort()
print(thislist)
```

```python
##sort the list numerically :
thislist=[1,100,15,20,60,5]
thislist.sort()
print(thislist)
```

CASE INSENSITIVE SORT 

 by default the sort() methosd is case sensitive , resulting in all capital letters being sorted before lower case letters :
 

```python
##case sensitive sorting can give an unexpected results 

thislist = ['apple','banana','Cherry','banana', 'Kiwi']
thislist.sort()
print(thislist)
```

```python
## perform a case sensitive sort oft he list :
## so if we want a case sensitive sort function , use str.lower as a key function .
thislist = ['apple','banana','Cherry','banana', 'Kiwi']
thislist.sort(key=str.lower)
print(thislist)
```

REVERSE ORDER 
The reverse() method reverses the current sorting order of the elements.


```python
##Reverse the order of the list item 
thislist = ['apple','banana','Cherry','banana', 'Kiwi']
thislist.reverse()
print(thislist)
```
